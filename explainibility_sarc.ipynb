```python
from transformers import RobertaTokenizer, RobertaModel

tokenizer = RobertaTokenizer.from_pretrained("roberta-base")

model = RobertaModel.from_pretrained(
    "roberta-base",
        output_attentions=True
        )
```

```python
result = []

# Extract attention scores from the last layer, averaging across heads
# We're taking the attention of the first token (usually <s> or [CLS]) to all other tokens.
scores = attention[-1][0, :, 0, :].mean(dim=0).tolist()

for token, score in zip(tokens, scores):
    if token not in ["<s>", "</s>", "<pad>"]:
            result.append((token, float(score)))

result.sort(key=lambda x: x[1], reverse=True)

print(result[:3])
```

```python
import pandas as pd

df = pd.read_json("Sarcasm_Headlines_Dataset.json", lines=True)

print(df.head())
```

```python
from transformers import RobertaTokenizer, RobertaForSequenceClassification

tokenizer = RobertaTokenizer.from_pretrained("roberta-base")

model = RobertaForSequenceClassification.from_pretrained(
    "roberta-base",
        num_labels=2
        )

print("RoBERTa loaded")
```

```python
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

print("Using:", device)
```

```python
from sklearn.model_selection import train_test_split

train_df, test_df = train_test_split(
    df[["headline", "is_sarcastic"]],
        train_size=3000,
            test_size=500,
                random_state=42,
                    stratify=df["is_sarcastic"]
                    )

print(len(train_df), len(test_df))
```

```python
train_tokens = tokenizer(
      train_df["headline"].tolist(),
          padding=True,
              truncation=True,
                  max_length=128,
                      return_tensors="pt"
                      )

train_labels = torch.tensor(
                          train_df["is_sarcastic"].tolist()
                          )

print("Tokenization done")
```

```python
from torch.utils.data import TensorDataset, DataLoader
from torch.optim import AdamW

dataset = TensorDataset(
    train_tokens["input_ids"],
        train_tokens["attention_mask"],
            train_labels
            )

loader = DataLoader(dataset, batch_size=16, shuffle=True)

optimizer = AdamW(model.parameters(), lr=2e-5)

model.train()

for batch in loader:
                input_ids, attention_mask, labels = [x.to(device) for x in batch]

optimizer.zero_grad()

output = model(
                                input_ids=input_ids,
                                        attention_mask=attention_mask,
                                                labels=labels
                                                    )

loss = output.loss
loss.backward()
optimizer.step()

print("Light fine-tuning finished!")
```

```python
model.eval()

text = "thirtysomething scientists unveil doomsday clock of hair loss"
actual = 1

inputs = tokenizer(text, return_tensors="pt").to(device)

with torch.no_grad():
    output = model(**inputs)

    prediction = output.logits.argmax(dim=1).item()

    print("Text:", text)
    print("Actual:", actual)
    print("Prediction:", prediction)
    print("Correctly Classified:",
          "Yes" if prediction == actual else "No")
```

```python
target_df = mustard_df[mustard_df["Sarcasm"].notna()].copy()

print(target_df.shape)
print(target_df[["KEY", "SENTENCE", "Sarcasm"]].head())
```

```python
print(mustard_df[mustard_df["KEY"].str.startswith("1_10004")][["KEY", "SENTENCE", "Sarcasm"]])
```

```python
# Get the 200 samples: 100 sarcastic + 100 non-sarcastic
sarcastic = target_df[target_df["Sarcasm"] == 1].sample(100, random_state=42)
non_sarcastic = target_df[target_df["Sarcasm"] == 0].sample(100, random_state=42)

samples = pd.concat([sarcastic, non_sarcastic]).sample(frac=1, random_state=42)

print(samples.shape)
```

```python
def get_context(key):
      context = mustard_df[
              mustard_df["KEY"].str.startswith(key.split("_u")[0] + "_c_")
                  ]["SENTENCE"].dropna().tolist()

      return "\n".join(context)

samples["Context"] = samples["KEY"].apply(get_context)

print(samples[["KEY", "Context", "SENTENCE", "Sarcasm"]].head())
```

```python
annotation_mustard = samples[
      ["KEY", "Context", "SENTENCE"]
      ].copy()

annotation_mustard.insert(
          0,
              "Sample ID",
                  ["M" + str(i).zfill(3) for i in range(1, 201)]
                  )

annotation_mustard = annotation_mustard.rename(columns={
                      "KEY": "MUSTARD ID",
                          "SENTENCE": "Target Sentence"
                          })

annotation_mustard["Annotator 1"] = ""
annotation_mustard["Annotator 2"] = ""
annotation_mustard["Annotator 3"] = ""
annotation_mustard["Annotator 4"] = ""

annotation_mustard.head()
```
